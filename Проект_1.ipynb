```python
def clean_numeric(value):
    if isinstance(value, str):  # Проверяем, является ли значение строкой
        value = value.replace("$", "").replace(",", "").strip()  # Убираем знак доллара, запятые и пробелы
        
        if value.endswith("M"):  # Если строка заканчивается на 'M' (миллионы)
            return float(value.replace("M", "")) * 1_000_000  # Удаляем 'M' и умножаем на 1 миллион
        
        elif value.endswith("K"):  # Если строка заканчивается на 'K' (тысячи)
            return float(value.replace("K", "")) * 1_000  # Удаляем 'K' и умножаем на 1000

    try:
        return float(value)  # Преобразуем оставшееся значение в число
    except ValueError:  # Если преобразование не удалось (например, строка содержит нечисловые данные)
        return np.nan  # Возвращаем NaN (значение отсутствует)
```

```python
df2["Revenue_Category"] = pd.qcut(df2["Revenue_$"], q=2, labels=["Низкая", "Высокая"])
df2
```

```python
# Преобразование даты
df2["Release_Date"] = pd.to_datetime(df2["Release_Date"], errors="coerce")

# Извлечение дня месяца и дня недели
df2["Day_of_Month"] = df2["Release_Date"].dt.day
df2["Day_of_Week"] = df2["Release_Date"].dt.dayofweek

# Определение будних и выходных дней (0-4 — будни, 5-6 — выходные)
df2["Weekday_Weekend"] = df2["Day_of_Week"].apply(lambda x: "Будни" if x < 5 else "Выходные")
```

```python
df2
```

```python
# Создаем таблицу сопряженности
contingency_table = pd.crosstab(df2["Revenue_Category"], df2["Weekday_Weekend"])
```

```python
# Выполняем тест хи-квадрат
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)
print(f"Хи-квадрат: {chi2}, p-значение: {p}")
```

```python
alpha = 0.05
if p < alpha:
    print("Гипотеза H₀ отвергается: существует статистически значимая связь между выручкой и днем выхода фильма.")
else:
    print("Гипотеза H₀ не отвергается: статистически значимой связи между выручкой и днем выхода фильма нет.")
```

```python
# Проверка нормальности остатков: тест Шапиро-Уилка
X = regression_df["Budget_USD"]
y = regression_df["Vote_Average"]
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
residuals = model.resid

shapiro_stat, shapiro_p = shapiro(residuals)
print(f"Shapiro-Wilk test: статистика = {shapiro_stat:.4f}, p-значение = {shapiro_p:.5f}")

if shapiro_p > 0.05:
    print("Остатки распределены нормально — условие линейной регрессии выполняется.")
else:
    print("Остатки не распределены нормально — возможны нарушения в линейной модели.")
```
